```python
from dotenv import load_dotenv
import openai

# .env 환경변수 파일 가져오기
load_dotenv()
     
```

```python
client = openai.OpenAI()
```

```python
def translate_text_gpt(text: str, src_lang: str, trg_lang: str) -> str:
    system_instruction = f"""
    너는 번역 앱이다. 입력된 문장 {text}을 {src_lang}에서 {trg_lang}로 자연스럽게 번역한다.
    번역된 텍스트만 출력하고 추가 설명이나 원문을 포함하지 않는다. 그리고 , 다른 질문 또는
    번역과 상관없는 내용은 '모릅니다.' 로 반드시 답변하기.
    """
    messages = [
        {"role": "system", "content": system_instruction},
        {"role": "user", "content": text},
    ]
    response = client.chat.completions.create(
        model="gpt-4.1-mini",  # 최신/가성비 모델
        messages=messages,
        temperature=0.2
    )
    return response.choices[0].message.content or ""
    # or 기준으로 앞의 값이 없을 때는 오류 대신 "" 빈 문자열(기본값) 처리
```

```python
text = input('번역할 내용을 입력하기')
src_lang = "한국어"
trg_lang = "영어"
translate_text_gpt(text,src_lang,trg_lang)
```

```python
# 이 함수는 translate_text_gpt 함수를 실행하기 전에 유효성 검사하는 목적
def gradio_translate(text: str, src_lang: str, trg_lang: str) -> str:
    if not text.strip():
        return "번역할 내용을 입력하세요."
    if src_lang == trg_lang:
        return "원본언어와 번역언어를 다르게 선택하세요."
    return translate_text_gpt(text,src_lang,trg_lang)
```

```python
import gradio as gr

with gr.Blocks() as demo:
    gr.Markdown('# 🧙‍♂️초간단 번역')
    text_input = gr.Textbox(lines=5, label="번역할 내용을 입력하세요.", placeholder="여기에 내용을 입력하세요.")
    src_lang_dropdown = gr.Dropdown(choices=["한국어","영어","일본어"], value="한국어", label="원본 언어를 선택하세요.")
    trg_lang_dropdown = gr.Dropdown(choices=["한국어","영어","일본어"], value="영어", label="번역할 언어를 선택하세요.")
    translate_button = gr.Button("번역")
    output_text = gr.Textbox(lines=5,label="번역 결과",placeholder="변역된 결과가 여기에 표시됩니다.", interactive=False)

    # 버튼이 클릭되면 실행 함수, 함수의 입력값, 리턴결과
    translate_button.click(
        fn=gradio_translate,
        inputs=[text_input, src_lang_dropdown, trg_lang_dropdown],
        outputs=output_text
    )
demo.launch(inline=False)
```
